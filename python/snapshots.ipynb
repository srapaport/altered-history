```python
import os
import re
import pandas as pd
import matplotlib.pyplot as plt
from utils_analysis import *
import numpy as np
import seaborn as sb
from collections import Counter
```

```python
df = pd.read_csv("../examples/snapshots_FULL.csv", delimiter=',')
```

```python
x = sorted(set(df['amount of snapshots']))
x
```

```python
all_count = df['amount of snapshots'].value_counts()
y = [all_count.get(snap) for snap in x]
plt.bar(x, y)
plt.xlabel('amount of snapshots')
plt.ylabel('amount of origins')
plt.show()
print(all_count.get(1))
```

```python
import math
#ranges = [(0,1), (1,5), (5,10), (10,50), (50,100), (100, 150), (200, math.inf)]
ranges = [(0,1), (1,2), (3,5), (5,10), (10,20), (20, 50), (50, math.inf)]
y = [0]*len(ranges) 
for snap in x:
    for i in range(len(ranges)):
        if snap > ranges[i][0] and snap <= ranges[i][1]:
            y[i] += all_count.get(snap)
            break
```

```python
new_x = ["] "+str(val[0])+":"+str(val[1])+"]" for val in ranges]
plt.bar(new_x, y)
plt.xlabel('amount of snapshots')
plt.ylabel('amount of origins')
plt.show()
```

```python
median_amount_snapshots = np.median(df['amount of snapshots'])
average_amount_snapshots = np.mean(df['amount of snapshots'])
print("Median:", median_amount_snapshots)
print("Average:", average_amount_snapshots)
```

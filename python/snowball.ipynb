```python
import os
import re
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sb
from collections import Counter
```

```python
prefix = "/infres/ir800/rapaport/results/FULL"
class_files = os.listdir(prefix)
df = pd.DataFrame()
for file in class_files:
    file_path = os.path.join(prefix, file)
    print("parsing file: ", file_path)
    if os.path.isfile(file_path) and file_path != "/infres/ir800/rapaport/results/FULL/checkpoints":
        df = pd.concat([df, pd.read_csv(prefix + "/" + file, delimiter=";")])
#df
```

```python
prefix = "/infres/ir800/rapaport/results/FULL/focus/classes"
class_files = os.listdir(prefix)
df_root = pd.DataFrame()
for file in class_files: 
    print("parsing file ", file)
    df_root = pd.concat([df_root, pd.read_csv(prefix + "/" + file, delimiter=";")])
#df_root
```

```python
pattern = r"^refs/heads/(main|master|dev|devel|develop|development)$"
df_filtered = df.loc[df['branch_name'].apply(lambda x: re.search(pattern, x) is not None)]
df_root_filtered = df_root.loc[df_root['branch_name'].apply(lambda x: re.search(pattern, x) is not None)]
print("amount of origins with altered commits: ", len(df_filtered['origin'].drop_duplicates()))
print("amount of origins with root altered commits: ", len(df_root_filtered['origin'].drop_duplicates()))
```

```python
origin_counts = df_filtered['origin'].value_counts().to_dict()
origin_counts_root = df_root_filtered['origin'].value_counts().to_dict()
```

```python
merged_counts = {key: (origin_counts.get(key, 0), origin_counts_root.get(key, 0)) for key in set(origin_counts) | set(origin_counts_root)}
```

```python
merged_counts.get('https://github.com/pfactum/pf-kernel')
```

```python
import json 
with open("snowball.json", "w") as f:
    f.write(json.dumps(merged_counts, indent=1))
```

```python
filtered_counts = {key: value for key, value in merged_counts.items() if value[1] > 0 and value[0]!=value[1]}
sorted_filtered_counts = dict(sorted(filtered_counts.items(), key=lambda item: item[1][0] - item[1][1], reverse=True))
sorted_filtered_counts 
```

```python
sum_value_1 = sum(value[1] for value in merged_counts.values())
sum_value_1
```

```python
whole_data = np.array([[n, r] for [n, r] in json.load(open("snowball.json")).values()])
data = whole_data[whole_data[:,1] > 0]

print(f"total number of origins: {whole_data.shape[0]}")
print(f"total number of altered commits: {whole_data[:,0].sum()}")
print(f"total number of manually altered commits: {whole_data[:,1].sum()}")

print(f"number of origins with manually altered commits: {data.shape[0]}")
print(f"number of manually altered commits: {data[:,1].sum()}")
print(f"number of altered commits: {data[:,0].sum()}")

with_snowball = data[data[:,1] < data[:,0]]

print(f"number of origin with snowball effects: {with_snowball.shape[0]}")
print(f"mean length of the snowball effects: {np.mean(with_snowball[:,0]/with_snowball[:,1])}")
print(f"median length of the snowball effects: {np.median(with_snowball[:,0]/with_snowball[:,1])}")
```
